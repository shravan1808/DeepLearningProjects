```python
y_raw = np.array([
    0, 1, 2, 0, 1, 2, 0, 1, 2, 0,
    1, 2, 0, 1, 2, 0, 1, 2, 0, 1,
    2, 0, 1, 2, 0, 1, 2, 0, 1, 2,
    0, 1, 2, 0, 1, 2, 0, 1, 2, 0,
    1, 2, 0, 1, 2, 0, 1, 2, 0, 1,
    2, 0, 1, 2, 0, 1, 2, 0, 1, 2
], dtype=np.int64)
```

```python
# Feature Scaling
X_mean = X_raw.mean(axis=0)
X_std = X_raw.std(axis=0) + 1e-8
X_norm = (X_raw - X_mean) / X_std
```

```python
device = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
X_tensor = torch.tensor(X_norm,dtype=torch.float32,device=device)
y_tensor = torch.tensor(y_raw,dtype=torch.long,device=device)

old_shape = X_tensor.shape
X_tensor = X_tensor.reshape(15,16)
new_shape = X_tensor.shape
sample_slice = X_tensor[0,:2]
```

```python
print("[TASK 1 SUCCESS] Tensor Operations Verified:")
print(f"Tensor Device: {X_tensor.device} | Tensor Shape: {old_shape}")
print(f"Reshaped Tensor Shape: {new_shape} | Sample Tensor Slice: {sample_slice}")
```

```python
x = torch.tensor([2.0, 3.0], requires_grad=True)
loss = torch.sum(x ** 2)
loss.backward()
print("[TASK 2 SUCCESS] Autograd Engine Verified:")
print(f"Scalar Function Loss Output: {loss}")
print(f"Computed Analytical Gradient (dL/dx): {x.grad}")
```

```python
from torch.utils.data import Dataset, DataLoader

X = torch.tensor(X_norm, dtype=torch.float32,device=device)
y = torch.tensor(y_raw, dtype=torch.long,device = device)

class CustomerDataset(Dataset):
  def __init__(self,X,y):
    self.X=X
    self.y=y

  def __len__(self):
    return len(self.X)

  def __getitem__(self, index):
     return self.X[index], self.y[index]

dataset = CustomerDataset(X,y)

train_loader = DataLoader(dataset,batch_size=16,shuffle=True)

for X_batch,y_batch in train_loader:
  X_batch = X_batch.to(device)
  y_batch = y_batch.to(device)
  break

print("[TASK 3 SUCCESS] PyTorch DataLoader Active:")
print(f"Total Dataset Samples: {len(dataset)} | Configured Batch Size: {X_batch.shape[0]}")
print(f"First Batch Features Shape: {X_batch.shape} | First Batch Labels Shape: {y_batch.shape}")
```

```python
import torch.nn as nn

class CustomerClassifier(nn.Module):
  def __init__(self,input_dim,output_dim):
    super(CustomerClassifier,self).__init__()
    self.network = nn.Sequential(
        nn.Linear(input_dim,16),
        nn.ReLU(),
        nn.Linear(16,8),
        nn.ReLU(),
        nn.Linear(8,output_dim)
    )

  def forward(self,X):
    return self.network(X)
```

```python
model = CustomerClassifier(4,3).to(device)
loss_fn = nn.CrossEntropyLoss()
optimizer = torch.optim.Adam(model.parameters(),lr=0.01)
print("[TASK 4 SUCCESS] PyTorch Neural Network Architecture Assembled:")
print(f"Model Device Placement: {X.device}")
print(f"Sequential Architecture: {model}")
```

```python
optimizer.zero_grad()
output = model(X_batch)
loss = loss_fn(output,y_batch)
loss.backward()
optimizer.step()
post_loss=loss_fn(model(X_batch),y_batch)
print("[TASK 5 SUCCESS] Single PyTorch Step Execution Complete:")
print(f"Step Pre-Loss: {loss:.4f} | Post-Step Loss: {post_loss:.4f}")
print("Gradient Calculated for Output Layer: True")
```

```python
loss_history = []
for epoch in range(100):
  epoch_loss = 0
  for X_batch,y_batch in train_loader:
    optimizer.zero_grad()
    output = model(X_batch)
    loss = loss_fn(output,y_batch)
    epoch_loss += loss.item()
    loss.backward()
    optimizer.step()
  avg_loss = epoch_loss / len(train_loader)
  loss_history.append(avg_loss)
```

```python
print("[TASK 6 SUCCESS] PyTorch Training Loop Progress:")

for epoch in [1, 25, 50, 75, 100]:
    print(
        f"Epoch {epoch:03d}/100 | "
        f"CrossEntropyLoss: {loss_history[epoch - 1]:.4f}"
    )
```

```python
checkpoint = {
    "model_state_dict":model.state_dict(),
    "optimizer_state_dict":optimizer.state_dict(),
    "epoch":100,
    "loss":loss_history[-1]
}

torch.save(checkpoint,"model_checkpoint.pt")
print("[TASK 7 SUCCESS] Checkpoint Serialized to Disk:")
print(f"Checkpoint File Saved: 'model_checkpoint.pt'")
print(f"Saved Epoch State: 100 | Saved Loss Value: {loss_history[-1]:.4f}")
```

```python
new_checkpoint = torch.load("model_checkpoint.pt")

loaded_model = CustomerClassifier(4,3).to(device)
loaded_model.load_state_dict(new_checkpoint['model_state_dict'])
original_output = model(X_batch)
loaded_output = loaded_model(X_batch)
print("[TASK 8 SUCCESS] Model State Dictionary Successfully Restored:")
print(f"Pre-Restoration Predictions Match Post-Restoration: {torch.allclose(original_output, loaded_output)}")
print(f"Model Ready for Interrupted Training Resumption.")
```

```python
model.eval()
with torch.no_grad():
  output = model(X)
  predicted =  torch.argmax(output,dim=1)
  correct = (predicted == y).sum().item()
  total = y.size(0)
  accuracy = (correct / total)*100
print("[TASK 9 SUCCESS] Batch Model Evaluation Complete:")
print(f"Total Dataset Accuracy: {accuracy:.2f}%")
print(f"Correct Matches: {correct} / {total} Samples")
```
